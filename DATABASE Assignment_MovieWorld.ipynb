Introduction

The film industry remains incredibly competitive on a global scale. The industry’s components, such as studios, distributors, and exhibitors, heavily rely on data-driven methods to forecast a film’s potential commercial and overall success. Within this project, four datasets covering film metadata, sales data including box-office and production-budget figures, consumer reviews, and expert critic reviews, sourced from metacritic.com and The Numbers.com, will be analyzed. The purpose of the project is to identify the determinants of box-office sales and to design, implement, and encapsulate a relational database that supports this analysis. 

To further build upon this brief, the following main research question (MRQ) is proposed:

“To what extent do the characteristics of consumer and expert critic reviews, including how much reviewers agree or disagree with each other, explain variation in box-office revenue beyond the effect of production budget?”
